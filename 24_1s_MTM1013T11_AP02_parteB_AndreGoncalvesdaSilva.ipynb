  $$Q=\frac{V}{t};$$

O volume pode ser descrito $V=Ad$, com $A$ a  área da seção transversal circular e $d$ a profundidade desse quantidade que se quer medir ([ver ilustração](https://cdn.kastatic.org/ka-perseus-images/0585b956c3af0284cb5fb0f72e16215286d2dcd1.png))[²], logo

$$Q=\frac{V}{t}=\frac{Ad}{t}=A\frac{d}{t}=Av_{fluido},$$

com $v_{fluido}=d/t$ a velocidade do fluido;

Sobre essa análise qualitativa do processo de medição da quantidade de fluido que passa pela seção transversal circular, pode-se montar formulações mais precisas para a quantificação dessa quantidade $Q$, como segue:

- Da disciplina de Cálculo B (ou II) problemas associados à quantificação do fluxo de um fluido através de uma superfície $S$ é dada pela integral de superfície: 
  $$\begin{aligned} \mbox{FLUXO} & =\iint_S \mathbf{v}(x,y,z)\cdot \mathbf{n} dS=  \iint_{D_{xz}} \left(v_x,v_y,v_z\right)\cdot \overbrace{\dfrac{\mathbf{N}}{||\mathbf{N}||}}^{\mathbf{n}}\underbrace{||\mathbf{N}|| dA}_{dS} \\ { } & = \iint_{D_{xz}} \left(v_x,v_y,v_z \right)\cdot (n_x,n_y,n_z) dA  = \iint_{D_{xz}} (v_xn_x+v_yn_y+v_zn_z) dA \end{aligned},$$ 
  cuja resolução dará a quantidade de fluido que passa pela secção por unidade de tempo;

- Considere o campo de velocidades do fluido dado por $\mathbf{v}(x,y,z)=\left(0,b\left(1-\tfrac{\sqrt{x^2+z^2}}{r_0}\right)^{1/7},0\right)$, com o centro do tubo de raio $r_0$  alinhado com o vetor normal $\mathbf{n}=(n_x,n_y,n_z)=(0,1,0)=\mathbf{j}=\frac{\mathbf{N}}{||\mathbf{N}||}$ a superfície de corte seccional descrita acima;

- Devido a simetria circular da seção transversal, emprega-se coordenadas polares para a descrição do campo de velocidades e no cálculo da quantidade de fluxo, ie,

   - $\mathbf{v}(x,y,z)=\left(0,b\left(1-\tfrac{\sqrt{x^2+z^2}}{r_0}\right)^{1/7},0\right)\equiv \left(0, b\left(1-\tfrac{r}{r_0}\right)^{1/7} ,0 \right) \Rightarrow v_y(r,\theta) = v(r) = b\left(1-\tfrac{r}{r_0}\right)^{1/7}$;
   
   - Conforme dados acima, segue que $$ \begin{aligned} Q & =\iint_{D_{xz}} v_y(x,z)dA, \: D_{xz}:=\{ (x,z) \in \mathbb{R}^2 | 0 \leq x^2+z^2 \leq r_0 \} \\  { } & \equiv \iint_{D_{r\theta}} v_y(r,\theta)dA, \: D_{r\theta}:=\{ (r,\theta) \in \mathbb{R}^2 | 0 < r < r_0 \wedge 0 \leq \theta < 2\pi  \} \: (x=r\cos{\theta}, z=r \operatorname{sen}{\theta})    \\ { } & =\int_{0}^{r_0} \int_{0}^{2\pi} v_y(r,\theta) rdr d\theta =  \int_{0}^{2\pi} d \theta  \int_{0}^{r_0} r v_y(r,\theta) dr = 2 \pi  \int_{0}^{r_0} r v_y(r,\theta) dr= 2 \pi  \int_{0}^{r_0} r v(r) dr\end{aligned}.$$

- Os itens acima, com a lista de de Dados aplicáveis:

  - $D_{xz}:=\{ (x,z) \in \mathbb{R}^2 | 0 \leq x^2+z^2 \leq r_0 \} \equiv D_{xz}:=\{ (x,z) \in \mathbb{R}^2 | 0 \leq x^2+z^2 \leq 1 \}$;

```python
f = lambda x: np.exp(-p*abs(x))*np.cos(-q*abs(x)) #usei o numpy para facilitar trabalhar com vetores
np.set_printoptions(precision=5)
xk = np.array([-2, -1, 0, 1, 2]) #fiz um array no np para calcular essa funcao mais rapidamente
fxk = f(xk)

print("xk: \t %.5f %.5f %.5f %.5f  %.5f"%(xk[0], xk[1], xk[2], xk[3], xk[4]))
print("f(xk): \t %.5f  %.5f %.5f %.5f %.5f"%(fxk[0], fxk[1], fxk[2], fxk[3], fxk[4]))
```

PVI obtido: $\left\{ \begin{aligned} {} & \negthickspace ay'=d+\frac{cx}{by}, \:  x \in [1,2]\\ {} & \negthickspace y(1)=2  \: (h=0,1) \end{aligned} \right. \Leftrightarrow \left\{ \begin{aligned} {} & \negthickspace (+2)y'=(+8)+\frac{(+4)x}{(+1)y}\\ {} & \negthickspace y(1)=2   \end{aligned} \right. $

```python
#código de newton cotes

def newtoncotes(f, a, b, n, m):
    d = [2., 6., 8., 90., 288., 840., 17280., 28350.]
    c = [1., 1., 4., 1., 3., 7., 32., 12., 19., 75., 50., 41., 216., 27., 272., 751., 3577., 1323., 2989., 28350., 989., 588., -928., 10496., -4540.]

    conderro = 0
    intgr = 0
    
    if n < 1 or n > 8:
        conderro += 1
    
    if m % n != 0:
        conderro += 2

    if conderro != 0:
        print("condERRO = %s"%conderro)
        return None

    p = trunc(0.25*(n*(n + 2) + (n%2)))
    h = (b - a)/m

    for i in range(m+1):
        x = a + i*h
        y = f(x)
        j = p + trunc(0.5*n - abs((i%n) - 0.5*n))
        k = 1 + trunc((n - (i%n))/n) - trunc((m - (i%m))/m)
        intgr += y*c[j]*k

    intgr *= n*h/d[n-1]
    
    return intgr
```

```python
#código de sistemas lineares
def LUdecomp(A):
    n = A.shape[0]
    m = A.shape[1]

    pivot = np.zeros(n)
    for i in range(n):
        pivot[i] = i+1

    det = 1
    for j in range(0,n-1):
        p = j
        Amax = abs(A[j,j])
        for k in range(j+1,n):
            if abs(A[k,j]) > Amax:
                Amax = abs(A[k,j])
                p = k
            
        if p != j:
            for k in range(0,m):
                t = A[j,k]
                A[j,k] = A[p,k]
                A[p,k] = t
            
            s = pivot[j]
            pivot[j] = pivot[p]
            pivot[p] = s
            det = -det
        
        det = det*A[j,j]
        if A[j,j] != 0:
            r = 1/A[j,j]
            for i in range(j+1,n):
                mult = A[i,j]*r
                A[i,j] = mult
                for k in range(j+1,m):
                    A[i,k] = A[i,k] - mult*A[j,k]

    det = det*A[n-1,n-1]
    return (A, det, pivot)

def getL(LU):
    n = LU.shape[0]
    m = LU.shape[1]

    L = np.zeros((n,m))
    for i in range(n):
        for j in range(m):
            if i == j:
                L[i,j] = 1
            elif i > j:
                L[i,j] = LU[i,j]
    
    return L

def getU(LU):
    n = LU.shape[0]
    m = LU.shape[1]

    U = np.zeros((n,m))
    for i in range(n):
        for j in range(m):
            if i <= j:
                U[i,j] = LU[i,j]
    
    return U

def solveL(L, pivot, b):
    n = L.shape[0]

    y = np.zeros(n)
    k = int(pivot[0])
    y[0] = b[k-1]
    for i in range(1,n):
        soma = 0
        for j in range(0,i):
            soma = soma + L[i,j]*y[j]
            
        k = int(pivot[i])
        #tive que fazer essa adaptação porque o algorítmo do material não é genérico o suficiente
        #para permitir que seja qualquer matrix triangular inferior
        y[i] = (b[k-1] - soma)/L[i,i]

    return y

def solveU(U, y):
    n = U.shape[0]
    x = np.zeros(n)

    x[n-1] = y[n-1]/U[n-1,n-1]
    for i in reversed(range(0,n-1)):
        soma = 0
        for j in range(i+1,n):
            soma = soma + U[i,j]*x[j]

        x[i] = (y[i] - soma)/U[i,i]

    return x

def solveLU(A, b):
    LU, det, pivot = LUdecomp(A)
    L = getL(LU)
    U = getU(LU)
    y = solveL(L, pivot, b)
    x = solveU(U, y)
    return x

def choleskydecomp(A):
    n = A.shape[0]
    
    conderro = 0
    det = 1
    r = 1
    t = 1
    soma = 0
    for j in range(0,n):
        soma = 0
        for k in range(0,j):
            soma = soma + A[j,k]**2

        t = A[j,j] - soma
        if t > 0:
            A[j,j] = math.sqrt(t)
            r = 1/A[j,j]
            det = det*t
        else:
            conderro = 1
            print("Erro: matriz não definida positiva")
            return (A, conderro, det)
        
        for i in range(j+1,n):
            soma = 0
            for k in range(0,j):
                soma = soma + A[i,k]*A[j,k]

            A[i,j] = (A[i,j] - soma)*r
            #algorítmo no material retornava uma matriz que não era triangular inferior (ele não mexia nos elementos acima da diagonal)
            A[j,i] = 0

    return (A, conderro, det)

def solvecholesky(A, b):
    G, conderro, det = choleskydecomp(A)
    #cholesky não faz pivoteamento, pivot vai ser só [1,2,3]
    pivot = np.arange(1, A.shape[0]+1)
    y = solveL(G, pivot, b)
    x = solveU(G.T, y)
    return x

def solvegauss_seidel(A, b, tol, itermax):
    n = A.shape[0]
    x = np.zeros(n)
    v = np.zeros(n)
    for i in range(0,n):
        r = 1/A[i,i]
        for j in range(0,n):
            if i != j:
                A[i,j] = A[i,j]*r
            
        #A[i,i] = 0
        b[i] = b[i]*r
        x[i] = b[i]
        
    iter = 0
    normarel = 1
    while iter <= itermax:
        iter = iter + 1
        for i in range(0,n):
            soma = 0
            for j in range(0,n):
                if i != j: 
                    soma = soma + A[i,j]*x[j]
            
            v[i] = x[i]
            x[i] = b[i] - soma

        normanum = 0    
        normaden = 0
        for i in range(0,n):
            t = abs(v[i] - x[i])
            normanum = normanum + t**2
            normaden = normaden + abs(v[i])**2

        normarel = math.sqrt(normanum/normaden)
        if normarel <= tol:
            conderro = 0
        else:
            conderro = 1
        

    return x, normarel, conderro
```

```python
#código de ajuste funcional
#não tem um algorítmo de ajuste funcional explícito na apostila, mas utilizei o que foi descrito no moodle
def discretenorm(u, v):
    if len(u) != len(v):
        raise ValueError("Vectors must have the same length")

    return sum([u[i]*v[i] for i in range(len(u))])

def discretepolynfit(x, y, n):
    M = np.zeros((n+1,n+1))
    b = np.zeros(n+1)
    for i in range(n+1):
        for j in range(n+1):
            M[i,j] = discretenorm(x**i, x**j)
        
        b[i] = discretenorm(y, x**i)

    #talvez tenha outros métodos mais eficientes de resolver esse sistema, mas como o LU e gauss compacto
    #ta funcionando bem aqui eu vou manter
    return solveLU(M, b)

def discreteexpfit(x, y):
    M = np.zeros((2,2))
    b = np.zeros(2)
    for i in range(2):
        for j in range(2):
            M[i,j] = discretenorm(x**i, x**j)
        
        b[i] = discretenorm(np.log(y), x**i)

    return solveLU(M, b)
```

Tem solução em série de potências truncada com simpy.

```python
from sympy import dsolve, Function, pprint, sin, cos, N
from sympy.abc import x
import sympy.plotting as splt

y = Function("y")
eq = -2.0*y(x).diff(x, 1) + 8.0 +4.0*x/y(x)
ysol = dsolve(eq, hint="1st_power_series", ics={y(1): 2}, n=8).rhs
print(ysol)

ysoln = lambda x: -3.0 - 0.75*(x - 1)**2 + 1.375*(x - 1)**3 - 2.890625*(x - 1)**4 + 6.5859375*(x - 1)**5 - 15.826171875*(x - 1)**6 + 39.5006975446429*(x - 1)**7 + 5.0*x
xrange = np.linspace(1, 2, 300)
plt.plot(xrange, ysoln(xrange), "r-", lw=2)
plt.xlabel("$x$")
plt.ylabel("$y(x)$")
plt.grid()
plt.legend(["Solução por séries de potência"])
plt.show()
```

```python
df = lambda x,y: 4.0 + 2.0*x/y
x0 = 1.0
fx0 = 2.0
h = 0.1
n = 10
xsol,ysol = rk4(df, x0, fx0, h, n)

xmesh = np.linspace(0.0, 2.2, 100)
ymesh = np.linspace(2, 8, 100)

Xl, Yl = np.meshgrid(xmesh, ymesh)
P = Xl/Xl
Q = 4.0 +2.0*Xl/(Yl)

plt.plot(xrange, ysoln(xrange), "r-", lw=2)
plt.plot(xsol, ysol, "bo--")

plt.streamplot(Xl, Yl, P, Q, density=1.5, color="gray", linewidth=0.25)

plt.xlabel("$x$")
plt.ylabel("$y(x)$")
plt.xlim(0.8, 2.2)
plt.ylim(2, 8)
plt.grid()
plt.legend(["Solução por séries de potência", "Solução por RK4"])
plt.show()
```
