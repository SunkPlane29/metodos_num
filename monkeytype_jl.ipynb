```julia
function LUdecomp(A)
    n = size(A,1)
    m = size(A,2)
    println(n)
    println(m)

    Pivot = zeros(n)
    for i = 1:n
        Pivot[i] = i
    end
    det = 1
    for j = 1:n-1
        p = j
        Amax = abs(A[j,j])
        for k = j+1:n
            if abs(A[k,j]) > Amax
                Amax = abs(A[k,j])
                p = k
            end
        end
        if p != j
            for k = 1:m
                t = A[j,k]
                A[j,k] = A[p,k]
                A[p,k] = t
            end
            s = Pivot[j]
            Pivot[j] = Pivot[p]
            Pivot[p] = s
            det = -det 
        end
        det = det*A[j,j]
        if A[j,j] != 0
            r = 1/A[j,j]
            for i = j+1:n
                Mult = A[i,j]*r
                A[i,j] = Mult
                for k = j+1:n
                    A[i,k] = A[i,k] - Mult*A[j,k]
                end
            end
        end
    end
    det = det*A[n,n]
    return A, Pivot, det
end
```

```julia
A = [1.0 1.0 3.0 ; 5.0 2.0 1.0 ; 3.0 5.0 1.0]
lu, P, det = LUdecomp(A)
display(lu)
println(LinearAlgebra.det(A))
println(LinearAlgebra.det(lu))
println(det)
```
